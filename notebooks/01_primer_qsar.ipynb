# Baseline QSAR con datos sintéticos

Este cuaderno ilustra una tubería de regresión y validación cruzada. Tanto los descriptores como la respuesta son artificiales; no representan moléculas, un endpoint toxicológico ni desempeño real.

```python
from pathlib import Path

import pandas as pd
from sklearn.linear_model import Ridge
from sklearn.model_selection import KFold, cross_validate
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
```

```python
data = pd.read_csv(Path("datasets") / "synthetic_qsar.csv")
features = ["molecular_weight", "logp", "hbd", "hba"]
target = "synthetic_activity"

X = data[features]
y = data[target]
data.head()
```

El escalado y el estimador se combinan en una tubería para que el escalador se ajuste únicamente con los datos de entrenamiento de cada fold. Esta validación cruzada solo demuestra que el ejemplo se ejecuta; los datos sintéticos no permiten concluir nada sobre sustancias reales.

```python
model = make_pipeline(StandardScaler(), Ridge(alpha=1.0))
cv = KFold(n_splits=4, shuffle=True, random_state=17)
scores = cross_validate(
    model,
    X,
    y,
    cv=cv,
    scoring="neg_root_mean_squared_error",
)
fold_rmse = -scores["test_score"]
print("RMSE por fold:", fold_rmse.round(3))
print("RMSE medio (solo datos sintéticos):", round(fold_rmse.mean(), 3))
```
